```python
from sklearn import datasets as dt
from sklearn import tree     as tr
from matplotlib import pyplot as plt
from sklearn import model_selection as ms
from sklearn import metrics as mt
from sklearn import datasets as ds
from sklearn import linear_model as lm
import pandas as pd
import numpy as np
from six import StringIO
```

```python
x_train = pd.read_csv('x_training.csv')
y_train = pd.read_csv('y_training.csv')
x_val = pd.read_csv('X_validation.csv')
y_val = pd.read_csv('y_validation.csv')
x_test = pd.read_csv('X_test.csv')
y_test = pd.read_csv('y_test.csv')
```

```python
y_train = y_train.values.ravel()
y_val = y_val.values.ravel()
y_test = y_test.values.ravel()
```

```python
#define
model = lm.LogisticRegression(
                                C=10,
                                solver='liblinear',
                                max_iter=100                             
)
#training
model.fit(x_train, y_train)
#performace
yhat_val = model.predict(x_val)
yhat_train = model.predict(x_train)
yhat_test = model.predict(x_test)
acc_val = mt.accuracy_score(y_val, yhat_val)
acc_train = mt.accuracy_score(y_train, yhat_train)
acc_test = mt.accuracy_score(y_test, yhat_test)
pre_train= mt.precision_score(y_train, yhat_train)
pre_val= mt.precision_score(y_val, yhat_val)
pre_test= mt.precision_score(y_test, yhat_test)
rec_train= mt.recall_score(y_train, yhat_train)
rec_val= mt.recall_score(y_val, yhat_val)
rec_test= mt.recall_score(y_test, yhat_test)
f1_train= mt.f1_score(y_train, yhat_train)
f1_val= mt.f1_score(y_val, yhat_val)
f1_test= mt.f1_score(y_test, yhat_test)
```

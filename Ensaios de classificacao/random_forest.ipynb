```python
from sklearn import datasets as dt
from sklearn import tree     as tr
from matplotlib import pyplot as plt
from sklearn import model_selection as ms
from sklearn import metrics as mt
from sklearn import datasets as ds
from sklearn import linear_model as lm
import pandas as pd
import numpy as np
from six import StringIO
from sklearn import ensemble as en
```

```python
x_train = pd.read_csv('x_training.csv')
y_train = pd.read_csv('y_training.csv')
x_val = pd.read_csv('X_validation.csv')
y_val = pd.read_csv('y_validation.csv')
x_test = pd.read_csv('X_test.csv')
y_test = pd.read_csv('y_test.csv')
```

```python
y_train = y_train.values.ravel()
y_val = y_val.values.ravel()
y_test = y_test.values.ravel()
```

```python
def avaliar(y_real, y_pred):
    accuracy = mt.accuracy_score(y_real, y_pred)
    precision = mt.precision_score(y_real, y_pred)
    recall = mt.recall_score(y_real, y_pred)
    f1 = mt.f1_score(y_real, y_pred)
    return accuracy, precision, recall, f1
```

```python
resultados = []

for max_depth in range(1, 32):      
    for n_estimators in [10, 50, 100]:  # três valores para número de árvores

        model_rf = en.RandomForestClassifier(
            max_depth=max_depth,
            n_estimators=n_estimators
        )

        model_rf.fit(x_train, y_train)

        # previsões
        pred_train = model_rf.predict(x_train)
        pred_val = model_rf.predict(x_val)
        pred_test = model_rf.predict(x_test)

        # métricas
        train_metrics = avaliar(y_train, pred_train)
        val_metrics = avaliar(y_val, pred_val)
        test_metrics = avaliar(y_test, pred_test)

        resultados.append([
            max_depth, n_estimators,
            *train_metrics,
            *val_metrics,
            *test_metrics
        ])
```

```python
colunas = [
    'max_depth', 'n_estimators',
    'train_acc', 'train_prec', 'train_rec', 'train_f1',
    'val_acc', 'val_prec', 'val_rec', 'val_f1',
    'test_acc', 'test_prec', 'test_rec', 'test_f1'
]

df_resultados = pd.DataFrame(resultados, columns=colunas)
df_resultados
```

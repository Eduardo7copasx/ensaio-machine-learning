```python
from sklearn import datasets as dt
from sklearn import tree     as tr
from matplotlib import pyplot as plt
from sklearn import model_selection as ms
from sklearn import metrics as mt
from sklearn import datasets as ds
from sklearn import linear_model as lm
import pandas as pd
import numpy as np
from six import StringIO
from sklearn.metrics import r2_score
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.preprocessing import PolynomialFeatures
```

```python
x_train = pd.read_csv('X_training.csv')
y_train = pd.read_csv('y_training.csv')
x_val = pd.read_csv('X_validation.csv')
y_val = pd.read_csv('y_val.csv')
x_test = pd.read_csv('X_test.csv')
y_test = pd.read_csv('y_test.csv')
```

```python
degree = 2
poly = PolynomialFeatures(degree=degree)

x_train_poly = poly.fit_transform(x_train)
x_val_poly = poly.fit_transform(x_val)
x_test_poly = poly.fit_transform(x_test)
```

```python
lasso_model = Lasso(
                    alpha=1.3,
                    max_iter=1000
)

#model fit (ajuste, treinamento)

lasso_model.fit(x_train_poly, y_train)

#previsao
y_predt = lasso_model.predict( x_train_poly)
y_predv = lasso_model.predict( x_val_poly)
y_predtt = lasso_model.predict( x_test_poly)
```

```python
print("R2 treino:", lasso_model.score(x_train_poly, y_train))
print("R2 validação:", lasso_model.score(x_val_poly, y_val))
print("R2 teste:", lasso_model.score(x_test_poly, y_test))
```

```python
mse_train = mt.mean_squared_error(y_train, y_predt)
mse_val = mt.mean_squared_error(y_val, y_predv)
mse_test = mt.mean_squared_error(y_test, y_predtt)

print("MSE treino:", mse_train)
print("MSE validação:", mse_val)
print("MSE teste:", mse_test)
```

```python
rmse_train = np.sqrt(mt.mean_squared_error(y_train, y_predt))
rmse_val = np.sqrt(mt.mean_squared_error(y_val, y_predv))
rmse_test = np.sqrt(mt.mean_squared_error(y_test, y_predtt))

print("RMSE treino:", rmse_train)
print("RMSE validação:", rmse_val)
print("RMSE teste:", rmse_test)
```

```python
mae_train = mt.mean_absolute_error(y_train, y_predt)
mae_val   = mt.mean_absolute_error(y_val, y_predv)
mae_test  = mt.mean_absolute_error(y_test, y_predtt)

print("MAE treino:", mae_train)
print("MAE validação:", mae_val)
print("MAE teste:", mae_test)
```

```python
mape_train = mt.mean_absolute_percentage_error(y_train, y_predt)
mape_val   = mt.mean_absolute_percentage_error(y_val, y_predv)
mape_test  = mt.mean_absolute_percentage_error(y_test, y_predtt)

print("MAPE treino:", mape_train)
print("MAPE validação:", mape_val)
print("MAPE teste:", mape_test)
```
